```python
import joule.api
import json
from joule.utilities import human_to_timestamp as h2ts
from joule.api.lumen_link import DisplayedDataStreamElement, DisplayedEventStream, Axis, IRange, SettingWithSize, SettingType, SettingWithAxis, BasicSetting
node = joule.api.get_node("localforge-dev")
```

```python
stream1 = await node.data_stream_get("/External/IMU/Magnetometer")
stream2 = await node.data_stream_get("/demo/random")
elem0 = DisplayedDataStreamElement(data_stream_element=stream2.elements[0],
                                   display_name='Custom',
                                   color="#00BF06") 
elem1 = DisplayedDataStreamElement(data_stream_element=stream2.elements[1],
                                   axis=Axis.RIGHT,
                                   color="#FFFF16") 

event_stream = await node.event_stream_get("/demo/events")
estream1 = DisplayedEventStream(event_stream=event_stream,
                                label=SettingWithSize(setting_type=SettingType.FIXED,value='Test',size=20),
                                marker=SettingWithSize(setting_type=SettingType.FIXED,value='o',size=8),
                                position=SettingWithAxis(setting_type=SettingType.FIXED,axis=Axis.LEFT,value=0),
                                height=BasicSetting(setting_type=SettingType.FIXED, value=1))
estream2 = DisplayedEventStream(event_stream=event_stream,
                                display_name='2nd Copy',
                                label=SettingWithSize(setting_type=SettingType.ATTRIBUTE,value='desc',size=12),
                                filter=[[['desc','is','B'],['count','eq',1]],[['count','gt',5]]])
link = await node.create_lumen_link("http://lambda:4200",
                             data_stream_elements=[elem0,elem1],
                             event_streams = [estream1,estream2],
                             left_y_bounds=IRange(min=-3,max=1), # currently has no effect
                             nav_time_bounds=IRange(min=h2ts("1 August 2025"),max=h2ts("1 September 2025")),
                             main_time_bounds=IRange(min=h2ts("04:00 28 August 2025"),max=h2ts("06:00 28 August 2025")))
print(link)
```
